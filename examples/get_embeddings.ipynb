# Extracting `GLACIER` Embeddings

Use this notebook to pull a pretrained `GLACIER` model from Hugging Face and extract molecular embeddings

```python
import torch 
from huggingface_hub import snapshot_download
import sys
repo_dir = snapshot_download(repo_id="glacier-hf/Glacier-100k-Mi")
sys.path.append(repo_dir)
```

## Initialize `GLACIER`

```python
from glacier_student import Glacier

model = Glacier.from_pretrained("glacier-hf/Glacier-100k-Mi") # Minimol as a teacher model
```

## Prepare input data

```python
from data.dataloader import SmilesMoleculeDataset, build_dataloader

# Example data
example_smiles = ["CC(=O)OC1=CC=CC=C1C(=O)O", "Cn1c(=O)c2c(ncn2C)n(C)c1=O"]

# Construct multimodal dataset
dataset = SmilesMoleculeDataset(smiles=example_smiles)

# Initialize dataloader
dataloader = build_dataloader(dataset, batch_size=1)
```

## Extract Embeddings

```python
# Set model to evaluation mode
model.eval()

# Get embeddings 
batch = next(iter(dataloader))
with torch.no_grad():
    embeddings = model(batch)

embeddings.shape
```
